# Notebook 01 — Data Exploration

Explore the ISOT datasets.

**Import the Necessary Libraries**

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

**Load the Dataset, Add a label to the Dataset and Combine both Dataset into one Dataframe**

```python

df_fake = pd.read_csv("../data/fake.csv")
df_true = pd.read_csv("../data/true.csv")

df_fake['label'] = 0  # Fake
df_true['label'] = 1  # Real

df_isot = pd.concat([df_fake, df_true], ignore_index=True)
print(df_isot.head(10))  # first 5 rows
```

**Class distribution of the Dataset**

```python

df_isot['label'].value_counts().plot(kind='bar')
df_isot['label'].value_counts()
#plt.title('Class Distribution')
#plt.show()
```

Show missing value count

```python
df_isot.isnull().sum()
```

save concatenated dataset

```python
df_isot.to_csv("../data/df_isot.csv", index=False)
```
